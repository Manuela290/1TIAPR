# 1 - Dados do problema:

```python
X =[
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
]

# 1 = liberar oferta, 0 = não liberar
y = [0, 0, 0, 1]

for i in range(len(X)):
    print("Cliente", i + 1)
    print("Pagamento em dia:", X[i][0])
    print("Alto engajamento:", X[i][1])
    print("Liberar oferta:", y[i])
```

# 2 - Função de ativação:

```python
def degrau(z): #Função de ativação degrau: devolve 1 se z >= 0, senão 0.
    if z >= 0:
        return 1
    else:
        return 0
```

# 3 - Treinamento do perceptron sem viés:

```python
pesos = [0.0, 0.0]
taxa = 0.1
epocas = 10

for epoca in range(epocas):
    erros_epoca = 0
    print(f"\nÉpoca {epoca + 1}")

    for i in range(len(X)):
        entrada = X[i]
        esperado = y[i]

        peso_antes = pesos.copy()

        soma = (
            entrada[0] * pesos[0]
            + entrada[1] * pesos[1]
        )

        previsto = degrau(soma)
        erro = esperado - previsto

        pesos[0] += taxa * erro * entrada[0]
        pesos[1] += taxa * erro * entrada[1]

        if erro != 0:
            erros_epoca += 1

        print(
            f"Entrada: {entrada}, "
            f"Esperado: {esperado}, "
            f"Previsto: {previsto}, "
            f"Erro: {erro}"
        )

        print("Pesos antes:", peso_antes)
        print("Pesos depois:", pesos)

    print("Erros da época:", erros_epoca)

    if erros_epoca == 0:
        break

print("\nPesos finais:", pesos)
```

# 4 - Treinamento do perceptron com viés :

```python
pesos = [0.0, 0.0]
bias = 0.0
taxa = 0.1
epocas = 10

for epoca in range(epocas):
    erros_epoca = 0

    print(f"\nÉpoca {epoca + 1}")

    for i in range(len(X)):
        entrada = X[i]
        esperado = y[i]

        peso_antes = pesos.copy()
        bias_antes = bias

        soma = (
            entrada[0] * pesos[0]
            + entrada[1] * pesos[1]
            + bias
        )

        previsto = degrau(soma)
        erro = esperado - previsto

        pesos[0] += taxa * erro * entrada[0]
        pesos[1] += taxa * erro * entrada[1]
        bias += taxa * erro

        if erro != 0:
            erros_epoca += 1

        print(
            f"Entrada: {entrada}, "
            f"Esperado: {esperado}, "
            f"Previsto: {previsto}, "
            f"Erro: {erro}"
        )

        print("Pesos antes:", peso_antes)
        print("Pesos depois:", pesos)
        print("Bias antes:", bias_antes)
        print("Bias depois:", bias)

    print("Erros da época:", erros_epoca)

    if erros_epoca == 0:
        break

print("\nPesos finais com viés:", pesos)
print("Bias final:", bias)
```

# 5 - Testar os resultados:

# CP 2 - Perguntas Respondidas:

# 1. O que é um Perceptron?

## Resposta:

O Perceptron é o modelo mais simples e fundamental de rede neural artificial de camada única, concebido por Frank Rosenblatt e inspirado no funcionamento de um neurônio biológico. Ele atua como um clasificador linear binário, projetado para receber um conjunto de entradas númericas, ponderá-las, realizar uma soma ponderada a um limiar (bias) e passar o resultado por uma função de ativação para gerar uma única saída binária (0 ou 1).

# 2. Quais são os principais elementos que compõem um Perceptron?

## Resposta:

A arquitetura básica de um Perceptron é composta por:



*   **Entradas(X):** Os dados ou características forncecidos ao modelo.
*   **Pesos(W):** Coeficientes que determinam a força, importância ou grau de influência de cada entrada sobre a decisão do neurônio.
*   **Viés(Bias):** Um parâmetro adicional que serve para ajustar o limite necessário para o neurônio disparar, funcionando como um fator de correção que ajuda o modelo a acertar casos onde as entradas são zeradas ou difíceis de separar apenas com os pesos.
*   **Soma Ponderada:** O somatório matemático dos produtos de cada entrada pelo seu respectivo peso, acrescido do bias .
*   **Função de Ativação:** A regra matemática (como a função degrau) que converte o valor contínuo da soma em uma resposta discreta final.

# 3. Qual a função dos pesos dentro do Perceptron?

## Resposta:

Os pesos têm o papel fundamental de quantificar a relevância de cada variável de entrada para a tomada de decisão do modelo. Durante o processo de treinamento, o algoritmo ajusta iterativamente esses valores com base nos erros cometidos; pesos maiores aumentam a influência de uma entrada específica, enquanto pesos próximos de zero tornam a entrada quase irrelevante para a predição.

# 4. Qual é a função da Função de Ativação?

## Resposta:

A Função de Ativação atua como um mecanismo de limiar (bias) e transformação, mapeando o resultado contínuo e muitas vezes ilimitado da soma ponderada para um formato de saída padronizado e interpretável (neste caso, binário: $0$ ou $1$). Sem ela, a rede neural se comportaria apenas como uma regressão linear comum, incapaz de resolver problemas de tomada de decisão categórica.

# 7. Explique por que esse comportamento ocorreu. Observe principalmente a entrada [0,0].

## Resposta:

Esse comportamento estagnado ocorreu por uma restrição matemática fundamental da arquitetura sem viés. Para a instância de entrada ``[0, 0]``, a soma ponderada resulta obrigatoriamente em zero, fazendo com que a função degrau (que ativa para valores $\ge 0$) preveja incorretamente a saída ``1`` quando o valor esperado era ``0``. Como as entradas são formadas por zeros, a fórmula de atualização dos pesos (``pesos += taxa * erro * entrada``) multiplica o gradiente por zero, impedindo que os pesos se modifiquem nessa amostra. Isso cria um ciclo vicioso onde o modelo oscila sem conseguir deslocar sua linha de separação para acomodar corretamente a classe negativa originada nas entradas nulas.

# 11. Para que serve o viés em um Perceptron? Explique com suas palavras.

## Resposta:

o viés é um parâmetro aprendido durante o treinamento que controla o quão fácil ou difícil é para o neurônio ativar, tornando o modelo mais flexível e capaz de resolver problemas que não podem ser separados por uma reta que passa pela origem. Exemplo: no treinamento, o perceptron chegou aos pesos [0,2; 0,1] e a um viés de aproximadamente -0,2. Isso significa que o neurônio só ativa quando a soma das entradas ponderadas supera 0,2. # 12. Qual foi a principal diferença entre o treinamento sem viés e o treinamento com viés?

## Resposta:

Nota-se que sem viés o perceptron não converge. A principal diferença seria que com viés o erro cai até zero e o treino termina. Sem viés o erro fica travado em 4 por época, e os pesos voltam ao mesmo ponto [0, 0] no fim de toda época. O modelo não aprende nada, só repete o mesmo ciclo.

# 13. O viés substitui os pesos? Explique

## Resposta:

O viés não substitui os pesos porque os dois existem juntos e fazem trabalhos diferentes. A conta do perceptron é:

soma = w1·x1 + w2·x2 + b

Os pesos (w1, w2) dizem o quanto cada entrada importa. Cada peso multiplica a sua entrada, então se a entrada for 0, aquele peso não contribui em nada.

O viés (b) é um valor que não depende de nenhuma entrada. Ele é somado no final, sempre, e funciona como o limiar que a soma precisa superar.

# 14. Um único Perceptron com viés seria suficiente para resolver um problema XOR? Explique.

## Resposta:

Um único Perceptron com viés não é suficiente para resolver o problema XOR porque ele funciona apenas como um classificador linear, sendo incapaz de separar dados. Geometricamente, as combinações do XOR (onde (0,0) e (1,1) resultam em 0, enquanto (0,1) e (1,0) resultam em 1) estão dispostas em cantos opostos de um gráfico 2D, tornando impossível que uma única linha reta isole os resultados verdadeiros dos falsos. Para superar essa limitação, é necessário utilizar uma rede neural mais avançada, como um Multilayer Perceptron (MLP), que empregue camadas ocultas e funções de ativação capazes de criar fronteiras de decisão complexas.

# 15. Em 3 a 6 linhas, explique o que você aprendeu sobre pesos, erro, épocas, convergência e viés.

## Respostas:

Pesos: Os pesos definem o quanto cada entrada influencia a decisão do perceptron e são ajustados a cada erro cometido.

Erro: O erro é a diferença entre a saída esperada e a prevista, e é ele que indica se e quanto os pesos devem mudar.

Época: A época é uma passagem completa por todos os exemplos de treino.

Convergência: A convergência acontece quando uma época termina com zero erros, como na época 4 do AND com viés.

Viés: O viés é um valor somado à soma ponderada que desloca a fronteira de decisão e dá ao neurônio um limiar ajustável: sem ele, o treino ficou preso em 4 erros por época e nunca convergiu. Mesmo com viés, um único perceptron só resolve problemas linearmente separáveis, por isso o AND funciona, mas o XOR exige uma rede com mais de um neurônio.
